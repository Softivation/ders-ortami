# İki GPS noktası arası mesafe ve yön

Görev sırasında bir sonraki waypoint'e ne kadar kaldığını ve hangi yöne dönüleceğini haversine formülüyle hesaplayın. Koordinatları kendi sahanızınkiyle değiştirin.

Kodu değiştirip hücreyi yeniden çalıştırın (Shift+Enter). Konunun tamamı: [softivation.com/egitimler/sifirdan-mission-planner-egitimi](https://softivation.com/egitimler/sifirdan-mission-planner-egitimi)

```python
from math import radians, degrees, sin, cos, atan2, sqrt

R = 6371000  # Dünya'nın ortalama yarıçapı (m)

def mesafe_yon(enlem1, boylam1, enlem2, boylam2):
    f1, f2 = radians(enlem1), radians(enlem2)
    df, dl = radians(enlem2 - enlem1), radians(boylam2 - boylam1)
    a = sin(df / 2) ** 2 + cos(f1) * cos(f2) * sin(dl / 2) ** 2
    mesafe = 2 * R * atan2(sqrt(a), sqrt(1 - a))
    yon = degrees(atan2(sin(dl) * cos(f2), cos(f1) * sin(f2) - sin(f1) * cos(f2) * cos(dl))) % 360
    return mesafe, yon

kalkis = (40.2307, 29.0099)     # örnek: Bursa
waypoint = (40.2335, 29.0152)

d, y = mesafe_yon(*kalkis, *waypoint)
print(f"Mesafe: {d:.1f} m | yön: {y:.0f} derece")
print(f"10 m/sn hızla yaklaşık {d / 10:.0f} saniye")
```

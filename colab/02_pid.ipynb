# PID ile irtifa tutma

Sanal bir multikopter 10 metreye çıkıp orada kalmaya çalışıyor. Kp, Ki ve Kd ile oynayın: aşımın, salınımın ve yerleşme süresinin nasıl değiştiğini grafikte görün.

Kodu değiştirip hücreyi yeniden çalıştırın (Shift+Enter). Konunun tamamı: [softivation.com/egitimler/sensorlerden-ucusa-temel-iha-egitimi](https://softivation.com/egitimler/sensorlerden-ucusa-temel-iha-egitimi)

```python
import matplotlib.pyplot as plt

kp, ki, kd = 1.2, 0.15, 0.9   # bu üç değerle oynayın
hedef = 10.0                  # metre
dt = 0.05                     # saniye

irtifa = hiz = integral = 0.0
onceki_hata = hedef
zaman, iz = [], []
for adim in range(400):
    hata = hedef - irtifa
    integral += hata * dt
    turev = (hata - onceki_hata) / dt
    onceki_hata = hata
    ivme = kp * hata + ki * integral + kd * turev
    ivme = max(-5.0, min(5.0, ivme))   # motorların sınırı: 5 m/sn2 (ivme)
    hiz = (hiz + ivme * dt) * 0.98     # hava direnci
    irtifa += hiz * dt
    zaman.append(adim * dt)
    iz.append(irtifa)

print(f"Son irtifa: {irtifa:.2f} m | en yüksek: {max(iz):.2f} m | aşım: {max(iz) - hedef:.2f} m")
plt.figure(figsize=(7, 3.2))
plt.plot(zaman, iz, label="irtifa")
plt.axhline(hedef, ls="--", c="gray", label="hedef")
plt.xlabel("zaman (sn)"); plt.ylabel("irtifa (m)"); plt.legend(); plt.title("PID ile irtifa tutma")
plt.show()
```
